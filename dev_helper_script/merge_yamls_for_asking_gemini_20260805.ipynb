```python
import glob
import os
import yaml


def merge_yaml_files(
    input_folder: str = ".", output_file: str = "merged_quizzes.yaml"
):
    """Finds all YAML files in input_folder, appends the filename under the key

    'yaml_file', wraps them in a list, and writes to output_file.
    """
    merged_data = []

    # Get all .yaml and .yml files in the folder (excluding the output file if present)
    pattern_yaml = os.path.join(input_folder, "*.yaml")
    pattern_yml = os.path.join(input_folder, "*.yml")
    yaml_files = sorted(
        set(glob.glob(pattern_yaml) + glob.glob(pattern_yml))
    )

    for file_path in yaml_files:
        filename = os.path.basename(file_path)

        # Skip the merged output file if running in the same directory
        if filename == output_file:
            continue

        with open(file_path, "r", encoding="utf-8") as f:
            data = yaml.safe_load(f)

            # Ensure data loaded as a dictionary
            if isinstance(data, dict):
                # Construct a new dict with yaml_file at the top
                merged_entry = {"yaml_file": filename}
                merged_entry.update(data)
                merged_data.append(merged_entry)

    # Save the consolidated list to the destination YAML file
    with open(output_file, "w", encoding="utf-8") as f:
        yaml.dump(
            merged_data,
            f,
            allow_unicode=True,
            sort_keys=False,
            default_flow_style=False,
        )

    print(
        f"Successfully merged {len(merged_data)} files into '{output_file}'."
    )


if __name__ == "__main__":
    merge_yaml_files(
        input_folder=r"C:\dev\temp\20260805_merge_yaml\Input",
        output_file=r"C:\dev\temp\20260805_merge_yaml\output.yaml"
    )
```
